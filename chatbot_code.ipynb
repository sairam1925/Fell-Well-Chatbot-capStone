# BuddyBot - Friendly, Comforting & Motivating Chatbot

A conversational AI companion built with **Python, NLTK, and Scikit-Learn** designed to detect emotional states, provide genuine comfort, crack funny jokes, and deliver motivational pep talks for students and everyday users.

### Step 1: Import Libraries

```python
import io
import random
import string
import warnings
import numpy as np
import nltk
from nltk.stem import WordNetLemmatizer
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

warnings.filterwarnings('ignore')
```

### Step 2: Download NLTK Datasets

```python
for res in ['punkt', 'punkt_tab', 'wordnet', 'omw-1.4']:
    nltk.download(res, quiet=True)
print("NLTK data packages loaded successfully.")
```

### Step 3: Load Motivational and Comfort Knowledge Base

```python
import os

corpus_file = 'corpus.txt'
if not os.path.exists(corpus_file):
    corpus_file = '../corpus.txt'

with open(corpus_file, 'r', encoding='utf-8', errors='ignore') as f:
    raw_text = f.read().lower()

sent_tokens = nltk.sent_tokenize(raw_text)
print(f"Corpus loaded successfully. Total sentences: {len(sent_tokens)}")
```

### Step 4: Emotion Recognition & Response Engine

```python
from chatbot import Chatbot

bot = Chatbot()
print("BuddyBot is ready!")
```

### Step 5: Test Conversational Scenarios

```python
test_scenarios = [
    "I feel so sad and lonely today",
    "I am stressed about my exam and final year project",
    "Tell me a joke",
    "Hype me up!"
]

for q in test_scenarios:
    print(f"User: {q}")
    print(f"Bot: {bot.respond(q)}\n")
```
